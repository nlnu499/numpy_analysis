```python
import numpy as np
```

print("numpy practice")

**rows are days and columns are products**


```python
data = np.array([
    [120, 150, 130],
    [100, 160, 120],
    [140, 170, 150],
    [130, 180, 160],
    [110, 140, 135],
    [150, 190, 175],
    [160, 200, 180]
])
```

```python
data
```

```python
print("Raw Data")
print(data)
```

```python
total_sales=np.sum(data)
```

```python
avg = np.mean(data)
```

```python
max= np.max(data)
```

```python
min= np.min(data)
```

```python
print("Sales", total_sales)
print("avg", avg)
print("max",max)
print("min",min)
```

** axis = 0 (column-wise) to find which product perform best and axis= 1 (row wise) to find which day has high earnings**

```python
product_totals = np.sum(data, axis=0)
product_avg = np.mean(data, axis=0)
```

```python
daily_totals = np.sum(data, axis=1)
```

```python
for i , total in enumerate(daily_totals):
    print(f"Day{i+1}:Total_sales={total}")
```

```python
print()
```

```python
normalized = (data - np.min(data)) / (np.max(data) - np.min(data))
```

```python
print("Normalized Data:")
print(normalized)
print()
```

```python
best_day = np.argmax(daily_totals) + 1
best_product = np.argmax(product_totals) + 1
```

```python
print(f"Best Sales Day: Day {best_day}")
print(f"Best Product: Product {best_product}")

print("End")
```
